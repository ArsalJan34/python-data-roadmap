```python
import numpy as np
```

```python
sales = np.array([
  [10,15,20,12,18,22,30],
  [5,8,6,9,12,10,11],
  [20,25,30,22,28,35,40],
  [2,3,2,4,5,3,4]
])
sales
```

```python
print("Total Sales: ", np.sum(sales))
```

```python
print("Total sales per day:",np.sum(sales,axis=0))
```

```python
print("Max Sales on a day: ",np.sum(sales,axis=0).max())
```

```python
print("Best selling day: ", np.argmax(np.sum(sales,axis=0))+1)
```
