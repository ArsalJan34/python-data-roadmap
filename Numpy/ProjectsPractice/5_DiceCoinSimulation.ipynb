```python
import numpy as np
```

```python
rolls = np.random.randint(1,7, size=1000)
```

```python
for i in range(1,7):
  count = np.count_nonzero(rolls == i)
  print(f"Number {i} came {count} times")
```

```python
counts = np.array([np.count_nonzero(rolls == i) for i in range(1,7)])
counts
```

```python
max_counts = np.max(counts)
most_frequent_num = np.argmax(counts) + 1
print(f"Max times came: {max_counts}")
print(f"Number frequently came: {most_frequent_num}")
```
