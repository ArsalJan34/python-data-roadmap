```python
import numpy as np
```

```python
marks = np.array([85,90,50,49,60,76,87,95])
marks
```

```python
print("Average: ",np.mean(marks))
```

```python
print("Max Marks : ",np.max(marks))
```

```python
print("Min Marks: ", np.min(marks))
```

```python
print("Passed: ", marks[marks >= 80])
```

```python
print("Sum Marks: ",np.sum(marks))
```
