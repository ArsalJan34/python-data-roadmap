```python
import numpy as np
```

```python
prices = np.array([100,250,80,400])
qty= np.array([2,1,3,4])

```

```python
total = prices * qty
print("Bill per item: ", total)
print("Grand total: ",np.sum(total))
```

```python
if np.sum(total) > 500:
  print(f"Before Discount: {np.sum(total)} \nAfter 10% discount: {np.sum(total) * 0.9}")
```
